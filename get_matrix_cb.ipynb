```python
def zero(m):
    n = m.shape[0]
    for i in range(n):
        m[i, i] = 0
    return m

def get_bar(m):
    m = m.ravel().tolist()

    x = []
    for i in range(20):
        x.append(i * 10)

    y = [0] * 20
    for e in m:
        if e > 200:
            y[19] += 1
        else:
            y[int(e // 10)] += 1

    return x, y

def get_bar_5(m):
    m = m.ravel().tolist()

    x = []
    for i in range(40):
        x.append(i * 5)

    y = [0] * 40
    for e in m:
        if e > 200:
            y[-1] += 1
        else:
            y[int(e // 5)] += 1

    return x, y

def get_bar_2(m):
    m = m.ravel().tolist()

    x = []
    for i in range(100):
        x.append(i * 2)

    y = [0] * 100
    for e in m:
        if e > 200:
            y[-1] += 1
        else:
            y[int(e // 2)] += 1

    return x, y

def get_bar_1(m):
    m = m.ravel().tolist()

    x = []
    for i in range(200):
        x.append(i * 1)

    y = [0] * 200
    for e in m:
        if e > 200:
            y[-1] += 1
        else:
            y[int(e // 1)] += 1

    return x, y

def get_bar_zero(m):
    m = zero(m)
    m = m.ravel().tolist()

    x = []
    for i in range(20):
        x.append(i * 10)

    y = [0] * 20
    for e in m:
        if e > 200:
            y[19] += 1
        else:
            y[int(e // 10)] += 1

    return x, y

def get_percent(l):
    total = sum(l)
    if total == 0:
        return [0] * len(l)
    return [round((e / total) * 100, 2) for e in l]
```
