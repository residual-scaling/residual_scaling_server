```python
import matplotlib.pyplot as plt
import numpy as np

# 数据准备
group_a = [11.7, 17.999999999999996, 21.1, 90.80000000000001]
group_b = [15.3, 21.8, 27.2, 34.4]
categories = ['Category 1', 'Category 2', 'Category 3', 'Category 4']

# 创建图形和坐标轴
plt.figure(figsize=(10, 6))

# 设置柱状图位置和宽度
x = np.arange(len(categories))
width = 0.35

# 绘制柱状图
bars_a = plt.bar(x - width/2, group_a, width, color='#1f77b4', label='Group A')
bars_b = plt.bar(x + width/2, group_b, width, color='#ff7f0e', label='Group B')

# 添加数据标签
plt.bar_label(bars_a, padding=3, fmt='%.1f')
plt.bar_label(bars_b, padding=3, fmt='%.1f')

# 设置图形属性
plt.ylabel('Values', fontsize=12)
plt.title('Comparison of Group A and Group B', fontsize=14, fontweight='bold')
plt.xticks(x, categories)
plt.legend()

# 添加网格线
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 调整y轴范围以容纳最高柱形
plt.ylim(0, max(max(group_a), max(group_b)) * 1.15)

# 添加数据来源说明
plt.figtext(0.5, 0.01, 'Source: Internal Data Analysis', 
            ha='center', fontsize=9, color='gray')

plt.tight_layout()
plt.show()
```
